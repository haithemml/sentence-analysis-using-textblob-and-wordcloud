```python
from transformers import pipeline
model = pipeline("text-generation", model="gpt2")
```

```python
sentence = model("Hi, My name is Peneloppe Criz, Iam here", do_sample=True, top_k=50,
                 temperature=0.9, max_length=100, num_return_sequences=2)
for i in sentence:
  print(i["generated_text"])
```
